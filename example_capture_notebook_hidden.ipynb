
# Example: Using `%%capture` in Jupyter (with export-friendly tags)

This notebook shows how to:
- Display normal outputs from a cell
- Execute a cell **silently** with `%%capture` (no output shown)
- Later **inspect or reveal** the captured output
- Hide **code inputs** on export with the `hide-input` tag

> To hide inputs in exported HTML, use **nbconvert** with tag removal enabled (see command below).


## A normal cell (visible output)

````python

import numpy as np
import matplotlib.pyplot as plt

print("This print is visible.")
x = np.linspace(0, 6*np.pi, 400)
y = np.sin(x) * np.exp(-0.05*x)
plt.plot(x, y)
plt.title("Damped Sine (visible)")
plt.show()
````


## A silent cell using `%%capture` (input hidden on export)

The next cell **runs but shows no output** (prints and plots are captured).  
It is tagged with **`hide-input`**, so when you export with nbconvert using the tag preprocessor,
**the code block itself will be hidden** in the resulting HTML.


````python

%%capture cap
import numpy as np
import matplotlib.pyplot as plt

print("This print is captured (not shown yet).")

x = np.linspace(0, 10, 200)
y = np.cos(x)
plt.figure()
plt.plot(x, y)
plt.title("Cosine (captured)")
plt.show()
````


Nothing displayed above — but the code **did run**.  
Now let's **inspect** what was captured.


````python

# Show captured standard output (prints)
print("Captured stdout:")
print(cap.stdout)

# Show captured standard error (if any)
print("\nCaptured stderr:")
print(cap.stderr)

# Reveal any rich display outputs (e.g., matplotlib figure) that were captured
cap.show()
````


## How to export with inputs hidden

Use **one** of the following commands, depending on your nbconvert version:

### Option A (tags-based, recommended)
```
jupyter nbconvert --to html --template pj  --TagRemovePreprocessor.enabled=True   --TagRemovePreprocessor.remove_input_tags hide-input   example_capture_notebook_hidden.ipynb
```

The "--template pj" part is to use the pretty jupyter template which makes the title work

THIS WORKS!!!


````python

# This cell will have its INPUT hidden on export if you use nbconvert with TagRemovePreprocessor
print("You'll see this output in the exported HTML, but not the code input.")
````


### Troubleshooting
- If you still see code inputs in the export, check your nbconvert version and try `--no-input` to verify the exporter is being used.
- Ensure the cell actually contains the `hide-input` tag (View → Cell Toolbar → Tags).  
- Some GUIs cache outputs; try exporting from the command line.
